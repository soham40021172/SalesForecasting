```python
cat_sales = (
    df.groupby('product_category_name_english')['order_item_id']
      .count()
      .sort_values(ascending=False)
)

cum_perc = (cat_sales.cumsum() / cat_sales.sum()) * 100
```

```python
threshold = 80
head_value = (cum_perc <= threshold).sum() + 1
print(f"The number of categories making up more than 80% is: {head_value}")
print(cum_perc.head(head_value))
```

```python
state_sales = (
    df.groupby('customer_state')['order_item_id']
      .count()
      .sort_values(ascending=False)
)

cum_st_perc = (state_sales.cumsum() / state_sales.sum()) * 100
```

```python
head_value_st = (cum_st_perc <= threshold).sum() + 1
print(f"The number of states making up more than 80% is: {head_value_st}")
cum_st_perc.head(head_value_st)
```

```python
top_categories = (
    df.groupby('product_category_name_english')['order_item_id']
    .count()
    .sort_values(ascending=False)
    .head(head_value)
)
top_categories
```

```python
top_states = (
    df.groupby('customer_state')['order_item_id']
    .count()
    .sort_values(ascending=False)
    .head(head_value_st)
)
top_categories
```

```python
top_categories = top_categories.index.tolist()
top_categories
```

```python
top_states = top_states.index.tolist()
top_states
```

```python
df['product_category_name_english'] = df['product_category_name_english'].apply(
    lambda x: x if x in top_categories else 'Other'
)
```

```python
df['customer_state'] = df['customer_state'].apply(
    lambda x: x if x in top_states else 'Other'
)
```

```python
daily_sales = df.groupby(
    ['date','product_category_name_english','customer_state']
).agg(
    sales=('order_item_id','count'),
    revenue=('price','sum')
).reset_index()
```

```python
daily_sales = daily_sales.rename(columns={
    "product_category_name_english": "category",
    "customer_state": "state"
})
```

```python
daily_sales = daily_sales.sort_values(
    ['date','state','category']
).reset_index(drop=True)
```

```python
full_date_range = pd.date_range(
    start=daily_sales['date'].min(),
    end=daily_sales['date'].max()
)
print(full_date_range)
```

```python
filled_list = []

for (category, state), group in daily_sales.groupby(['category','state']):
    
    group = group.copy()
    group['date'] = pd.to_datetime(group['date'])   # 🔥 ensure match
    group = group.set_index('date')
    
    group = group.reindex(full_date_range)
    
    group['category'] = category
    group['state'] = state
    group['sales'] = group['sales'].fillna(0)
    group['revenue'] = group['revenue'].fillna(0)
    
    group = group.reset_index().rename(columns={'index': 'date'})
    
    filled_list.append(group)

daily_sales_filled = pd.concat(filled_list, ignore_index=True)
```

```python
daily_sales_filled = daily_sales_filled.sort_values(
    ['category','state','date']
).reset_index(drop=True)
```
